```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import plotly.express as px
```

```python
os.chdir('../')
parent_dir = os.getcwd()
directory = "data\\raw"
data_path = os.path.join(parent_dir, directory)
```

```python
train_data_path = os.path.join(data_path, "train_data.csv")
train_labels_path = os.path.join(data_path, "train_labels.csv")
```

```python
column_lables = [str(i) for i in range(264)]
```

```python
db_train = pd.read_csv(train_data_path,index_col=False,names=column_lables)
```

```python
db_train.head()
```

```python
db_train_lable = pd.read_csv(train_labels_path,names=["label"])
```

```python
# Correlation threshold
threshold = 0.90

# Absolute correlation matrix
corr_matrix = db_train.corr().abs()

# Upper triangle only
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Features to remove
redundant_features = [
    column for column in upper.columns
    if any(upper[column] > threshold)
]
```

```python
# Remove constant and redundant features
features_to_remove = ['216', '217', '218', '219'] + redundant_features

db_train_clean = db_train.drop(columns=features_to_remove)

# Create processed folder
processed_path = os.path.join(parent_dir, "data", "processed")
os.makedirs(processed_path, exist_ok=True)

# Save cleaned data
db_train_clean.to_csv(
    os.path.join(processed_path, "train_features_clean.csv"),
    index=False
)

db_train_lable.to_csv(
    os.path.join(processed_path, "train_labels_clean.csv"),
    index=False
)
```

```python
metadata = pd.DataFrame({
    "Removed_Features": features_to_remove
})

metadata.to_csv(
    os.path.join(processed_path, "preprocessing_metadata.csv"),
    index=False
)
```
